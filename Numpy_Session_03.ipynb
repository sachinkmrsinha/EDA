**Cosine Similarity**

```python
x=[1,2,3]
y=[4,5,6]
```

```python
# Convert into array
# Nr=np.dot()
# dr = a*b
# a = norm(arr1)
```

```python
arr1 = np.array(x)
arr2 = np.array(y)
```

```python
Nr = np.dot(arr1,arr2)
norm_A= norm(arr1)
norm_B= norm(arr2)
Dr = norm_A*norm_B
Cosine_sim = Nr/Dr
Cosine_sim
```
