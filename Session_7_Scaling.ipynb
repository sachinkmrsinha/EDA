**Z score**

 `z = (x-u)/sigma`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
file_path=r"C:\Users\learn\Documents\NareshIT\EDA\Pandas\Visadataset - Visadataset.csv"
visa_df = pd.read_csv(file_path)
visa_df
```

```python
cat=visa_df.select_dtypes(include='object').columns
num=visa_df.select_dtypes(exclude='object').columns
```

```python
# Step-1: take the pwage data
# Step-2: calculate mean of the data
# Step-3: calculate std of the data
# Step-4: Nr=data-mean step1-step2
# Step-5: z_wage=Nr/std = Step4/Step3
# Step-6: Create a new column visa_df['z_wage'] = z_wage

pwage_data = visa_df['prevailing_wage']
mean = pwage_data.mean()
std = pwage_data.std()
Nr = pwage_data-mean
z_wage = Nr/std
visa_df['z_wage'] = z_wage
visa_df
```

```python
plt.figure(figsize=(8,4))
plt.subplot(2,2,1).hist(pwage_data)
plt.subplot(2,2,2).hist(z_wage)
# plt.subplot(2,2,3)
# plt.subplot(2,2,4)
```

```python
plt.figure(figsize=(8,4))
plt.subplot(2,2,1).hist(visa_df['prevailing_wage'])
plt.subplot(2,2,2).hist(visa_df['z_wage'])
```

```python
plt.figure(figsize=(8,4))
plt.suptitle("Raw-data vs Z-data")
plt.subplot(2,2,1).hist(pwage_data)
plt.title("prevailing_wage")
plt.subplot(2,2,2).hist(z_wage)
plt.title("z score of prevailing_wage")
plt.subplot(2,2,3).boxplot(pwage_data,vert=False)
plt.subplot(2,2,4).boxplot(z_wage,vert=False)
plt.show()
```

**Min Max Scalar**

```python
# Step-1: read the data
# Step-2: calculate min
# Step-3: calculate max
# Step-4: Nr = step1-step2
# Step-5: Dr = step3-step2
# Step-6: Nr/Dr
# Step-7: visa_df['wage_min_max'] = Nr / Dr
```

```python
pwage_data = visa_df['prevailing_wage']
pwage_min = pwage_data.min()
pwage_max = pwage_data.max()
visa_df['wage_min_max'] = (pwage_data - pwage_min) / (pwage_max - pwage_min)
```

```python
(pwage_data - pwage_min), (pwage_max - pwage_min)
```
