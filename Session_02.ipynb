# 3D

```python
l1=[10,20,30,40]
l2=[1,2,3,4]
l3=[100,200,300,400]
l=[l1,l2,l3]
#np.array(object,dtype)
arr3= np.array(l) # l==> object
arr3
```

```python
print("Size: ",arr3.size)
print("shape: ",arr3.shape)
print("dimesion: ",arr3.ndim)
```

```python
l1=[10,20,30,40]
l2=[1,2,3,4]
l3=['A','B','C','D']
l=[l1,l2,l3]
#np.array(object,dtype)
arr4= np.array(l) # l==> object
arr4
```

```python
l1=[10,20,30,40]
l2=[1,2,3,4]
l3=['A','B','C','D']
l=[[l1,l2,l3]]
#np.array(object,dtype)
arr5= np.array(l) # l==> object
arr5
```
