```python
# Step-1: Create a Folder
folder_name = 'Frequency_table'
os.makedirs(folder_name,exist_ok=True)

# Step-2: make a new path
directory=os.getcwd()
new_dir=os.path.join(directory,folder_name)

# Step-3: Create a DataFrame 
for i in cat[1:]:
    keys = visa_df[i].value_counts().keys()
    values = visa_df[i].value_counts().values
    df = pd.DataFrame(zip(keys,values),columns=['column','frequency'])
    
    # Step-4: Make the new path
    file_name=f'{i}_.csv'
    path=os.path.join(new_dir,file_name)
    
    # Step-5: Save into data frame
    df.to_csv(path,index=False)
```

**Saving Bar Chart**

```python
plt.figure(figsize=(10,4)) # x axis: 10 units width & y axis:4 units height
# Step-1: Create a Folder
folder_name = 'Bar_chart'
os.makedirs(folder_name,exist_ok=True)

# Step-2: make a new path
directory=os.getcwd()
new_dir=os.path.join(directory,folder_name)

# Step-3: Create a DataFrame 
for i in cat[1:]:
    keys = visa_df[i].value_counts().keys()
    values = visa_df[i].value_counts().values
    # df = pd.DataFrame(zip(keys,values),columns=['column','frequency'])

    col=['blue','pink','orange','purple','yellow','green']
    plt.bar(keys,values,width=0.5,color=col)
    plt.xlabel('Classes')
    plt.ylabel('Frequency')
    plt.title(f'{i}_Chart')
    
    # Step-4: Make the new path
    file_name=f'{i}_.jpg'
    path=os.path.join(new_dir,file_name)
    
    # Step-5: Save into data frame
    # df.to_csv(path,index=False)
    plt.savefig(path)
    plt.show()  

  
```

**Saving Pie Chart**

```python
plt.figure(figsize=(10,4)) # x axis: 10 units width & y axis:4 units height
# Step-1: Create a Folder
folder_name = 'Pie_chart'
os.makedirs(folder_name,exist_ok=True)

# Step-2: make a new path
directory=os.getcwd()
new_dir=os.path.join(directory,folder_name)

# Step-3: Create a DataFrame 
for i in cat[1:]:
    keys = visa_df[i].value_counts().keys()
    values = visa_df[i].value_counts().values
    # df = pd.DataFrame(zip(keys,values),columns=['column','frequency'])

    plt.pie(values,labels=keys,autopct='%1.2f%%',startangle=60,radius=0.6)
    plt.title(f'{i}_Chart')
    
    # Step-4: Make the new path
    file_name=f'{i}_.jpg'
    path=os.path.join(new_dir,file_name)
    
    # Step-5: Save into data frame
    # df.to_csv(path,index=False)
    plt.savefig(path)
    plt.show()  
    

  
```
