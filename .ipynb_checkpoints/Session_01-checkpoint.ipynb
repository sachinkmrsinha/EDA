**inplace**

- When we do the operation the result output is not overwritting 
- means the original dataframe is same as it is.
- in order to this we need to use **inplace**
- by default inplace= False
- make **inplace=True** if yoy want to overwrite the result

```python
d={'Names':['Ramesh','Suresh','Satish'],
  'Age':[20,24,28],
  'City':['Hyd','Pune','Patna'],
  'Marks': [100,200,300]}
df=pd.DataFrame(d)
df
```

```python
df.drop(columns='Marks') #Multiple column the use list
```

```python
df.drop(columns='Marks',axis=1,inplace=True) 
```

```python
df
```

**Step-14: Rename the column**

```python
d={'Names':['Ramesh','Suresh','Satish'],
  'Age':[20,24,28],
  'City':['Hyd','Pune','Patna'],
  'Marks': [100,200,300]}
df=pd.DataFrame(d)
df
```

```python
d={'City':'city'} # old:new
df.rename(d,axis=1,inplace="True")
```

```python
d={2:'B'} # old:new
df.rename(d,axis=0,inplace="True")
```
